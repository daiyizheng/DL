# 欢迎使用 FastAPI-MCP！
使用 Auth 将您的 FastAPI 端点公开为模型上下文协议 (MCP) 工具！

MCP（模型上下文协议）是定义 AI 代理如何与应用程序通信的新兴标准。使用 FastAPI-MCP，只需三行代码即可为您的应用程序创建安全的 MCP 服务器：

```Python
from fastapi import FastAPI
from fastapi_mcp import FastApiMCP

app = FastAPI()

mcp = FastApiMCP(app)
mcp.mount()
```


就这样！您自动生成的 MCP 服务器现在可以在https://app.base.url/mcp

​
## 特征
- 内置身份验证，使用您现有的 FastAPI 依赖项！
- FastAPI-native：不仅仅是另一个 OpenAPI -> MCP 转换器
- 无需任何配置- 只需将其指向您的 FastAPI 应用程序即可工作
- 保留请求模型和响应模型的模式
- 保存所有端点的文档，就像在 Swagger 中一样
- 灵活部署- 将您的 MCP 服务器安装到同一应用程序，或单独部署
- ASGI 传输- 直接使用 FastAPI 的 ASGI 接口进行高效通信

## 创建基本的 MCP 服务器
要创建基本的 MCP 服务器，请导入或创建一个 FastAPI 应用程序，用FastApiMCP类包装它并将 MCP 挂载到现有应用程序中：

```python
from fastapi import FastAPI
from fastapi_mcp import FastApiMCP

# Create (or import) a FastAPI app
app = FastAPI()

# Create an MCP server based on this app
mcp = FastApiMCP(app)

# Mount the MCP server directly to your app
mcp.mount()
if __name__ == "__main__":
    import uvicorn
    uvicorn.run(app, host="0.0.0.0", port=8000)
```

通过运行您的 FastAPI，您的 MCP 将以 运行https://app.base.url/mcp。

例如，通过使用 uvicorn

## 将客户端连接到 MCP 服务器
一旦您的带有 MCP 集成的 FastAPI 应用程序运行，您就需要将其连接到 MCP 客户端。

### 使用 SSE 连接到 MCP 服务器
对于任何支持 SSE 的 MCP 客户端，您只需提供 MCP url。

所有最流行的 MCP 客户端（Claude Desktop、Cursor 和 Windsurf）都使用以下配置格式：

```json
{
  "mcpServers": {
    "fastapi-mcp": {
      "url": "http://localhost:8000/mcp"
    }
  }
}
````

### 使用mcp-remote连接到 MCP 服务器
如果您想要支持身份验证，或者您的 MCP 客户端不支持 SSE，我们建议使用mcp-remote作为桥梁。

```json
{
  "mcpServers": {
    "fastapi-mcp": {
      "command": "npx",
      "args": [
        "mcp-remote",
        "http://localhost:8000/mcp",
        "8080"  // Optional port number. Necessary if you want your OAuth to work and you don't have dynamic client registration.
      ]
    }
  }
}
```

## 最佳实践
本指南概述了将标准 API 转换为模型上下文协议 (MCP) 工具以供 AI 代理使用的最佳实践。合理的工具设计有助于确保 LLM 能够理解并安全地使用您的 API。遵循这些最佳实践，您可以构建更安全、更直观的 MCP 工具，从而增强 LLM 代理的功能。

​
### 工具选择
- 精挑细选：避免将每个端点都暴露为工具。LLM 客户端使用数量有限且定义明确的工具会获得更好的性能，而提供商通常会对工具设置限制。

- 优先考虑安全性：除非绝对必要，否则请勿暴露或端点。LLM 具有不确定性，可能会无意中更改或损坏系统或数据库。PUTDELETE

- 专注于数据检索： 优先选择GET能够安全且可预测地返回数据的端点。

- 强调有意义的工作流程： 展现清晰、目标明确的任务。操作明确的工具更易于客服人员理解和有效使用。

​
### 工具命名
- 使用简短、描述性的名称：帮助 LLM 选择和使用正确的工具。请注意，某些 MCP 客户端会限制工具名称的长度。
- 遵循命名约束：
必须以字母开头只能包含字母、数字和下划线 避免使用连字符（例如，AWS Nova 不支持连字符）在所有工具中一致使用camelCase或snake_case
- 确保唯一性：每个工具名称都应该是唯一的，并清楚地表明其功能。

​
### 文档
- 有意义地描述每个工具：提供每个工具功能的清晰简洁的摘要。

- 包括使用示例和参数描述：这些有助于 LLM 了解如何正确使用该工具。

- 跨工具标准化文档：保持格式和结构的一致性以保证质量和可读性。
