# 工具命名
FastAPI-MCP 使用 `FastAPI` 路由中的 `operation_id` 作为 MCP 工具名称。如果您未指定`operation_id`，FastAPI 会自动生成一个，但这些名称可能比较隐晦。

比较这两个端点定义：

```python
# Auto-generated operation_id (something like "read_user_users__user_id__get")
@app.get("/users/{user_id}")
async def read_user(user_id: int):
    return {"user_id": user_id}

# Explicit operation_id (tool will be named "get_user_info")
@app.get("/users/{user_id}", operation_id="get_user_info")
async def read_user(user_id: int):
    return {"user_id": user_id}
```

为了获得更清晰、更直观的工具名称，我们建议operation_id在 FastAPI 路由定义中添加明确的参数。


## 自定义
### 服务器元数据
您可以通过修改以下内容来定义 MCP 服务器名称和描述：

```python
from fastapi import FastAPI
from fastapi_mcp import FastApiMCP

app = FastAPI()

mcp = FastApiMCP(
    app,
    name="My API MCP",
    description="Very cool MCP server",
)
mcp.mount()
```

### 工具和架构描述
创建 MCP 服务器时，可以通过更改 `describe_all_responses` 标志，在工具描述中包含所有可能的响应模式，或者通过更改 `describe_full_response_schema` 标志，在工具描述中包含完整的 JSON 模式。

```python
from fastapi import FastAPI
from fastapi_mcp import FastApiMCP

app = FastAPI()

mcp = FastApiMCP(
    app,
    name="My API MCP",
    description="Very cool MCP server",
    describe_all_responses=True,
    describe_full_response_schema=True
)

mcp.mount()
```

### 自定义暴露的端点
您可以使用 Open API 操作 ID 或标签来控制哪些 FastAPI 端点作为 MCP 工具公开，以便：

- 仅包含特定操作
- 排除特定操作
- 仅包含具有特定标签的操作
- 排除带有特定标签的操作
- 合并操作 ID 和标签

这些配置的相关参数是include_operations，exclude_operations，include_tags，exclude_tags，并且可以按如下方式使用：

```python
from fastapi import FastAPI
from fastapi_mcp import FastApiMCP

app = FastAPI()

mcp = FastApiMCP(
    app,
    include_operations=["get_user", "create_user"]
)
mcp.mount()
```

```python
from fastapi import FastAPI
from fastapi_mcp import FastApiMCP

app = FastAPI()

mcp = FastApiMCP(
    app,
    exclude_operations=["delete_user"]
)
mcp.mount()
```

```python
from fastapi import FastAPI
from fastapi_mcp import FastApiMCP

app = FastAPI()

mcp = FastApiMCP(
    app,
    include_tags=["users", "public"]
)
mcp.mount()
```

```python
from fastapi import FastAPI
from fastapi_mcp import FastApiMCP

app = FastAPI()

mcp = FastApiMCP(
    app,
    exclude_tags=["admin", "internal"]
)
mcp.mount()
```

```python
from fastapi import FastAPI
from fastapi_mcp import FastApiMCP

app = FastAPI()

mcp = FastApiMCP(
    app,
    include_operations=["user_login"],
    include_tags=["public"]
)
mcp.mount()
```

## 过滤注意事项
- 你不能同时`include_operations`使用`exclude_operations`
- 你不能同时`include_tags`使用`exclude_tags`
- 您可以将操作过滤与标签过滤结合起来（例如，`include_operations`使用`include_tags`）
- 组合筛选器时，将采用贪婪方法。符合任一条件的端点将被纳入
