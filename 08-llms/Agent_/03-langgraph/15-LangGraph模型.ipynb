# 模型
LangGraph通过 LangChain 库内置了对LLM（语言模型）的支持。这使得您可以轻松地将各种 LLM 集成到您的代理和工作流中。

## 初始化模型¶
用于init_chat_model初始化模型：

`pip install -U "langchain[openai]"`

```python
import os
from langchain.chat_models import init_chat_model

os.environ["OPENAI_API_KEY"] = "sk-..."

llm = init_chat_model("openai:gpt-4.1")
```

## 直接实例化模型¶
如果无法通过 获得模型提供程序`init_chat_model`，您可以直接实例化提供程序的模型类。该模型必须实现BaseChatModel 接口并支持工具调用：



```python
# Anthropic is already supported by `init_chat_model`,
# but you can also instantiate it directly.
from langchain_anthropic import ChatAnthropic

model = ChatAnthropic(
  model="claude-3-7-sonnet-latest",
  temperature=0,
  max_tokens=2048
)
```

## 在代理中使用¶
使用时，`create_react_agent`您可以通过名称字符串指定模型，这是使用 初始化模型的简写`init_chat_model`。这样您就可以使用该模型，而无需直接导入或实例化它。

```python
from langgraph.prebuilt import create_react_agent

create_react_agent(
   model="anthropic:claude-3-7-sonnet-latest",
   # other parameters
)
```

## 高级模型配置
### 禁用流媒体¶
要禁用单个 LLM 令牌的流式传输，请`disable_streaming=True`在初始化模型时设置：

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    "anthropic:claude-3-7-sonnet-latest",
    disable_streaming=True
)
```

### 添加模型回退¶

您可以使用以下方式添加对不同模型或不同 LLM 提供程序的回退model.with_fallbacks([...])：

```python
from langchain.chat_models import init_chat_model

model_with_fallbacks = (
    init_chat_model("anthropic:claude-3-5-haiku-latest")
    .with_fallbacks([
        init_chat_model("openai:gpt-4.1-mini"),
    ])
)
```

### 使用内置速率限制器¶
Langchain 内置内存速率限制器。该速率限制器是线程安全的，可由同一进程中的多个线程共享。

```python
from langchain_core.rate_limiters import InMemoryRateLimiter
from langchain_anthropic import ChatAnthropic

rate_limiter = InMemoryRateLimiter(
    requests_per_second=0.1,  # <-- Super slow! We can only make a request once every 10 seconds!!
    check_every_n_seconds=0.1,  # Wake up every 100 ms to check whether allowed to make a request,
    max_bucket_size=10,  # Controls the maximum burst size.
)

model = ChatAnthropic(
   model_name="claude-3-opus-20240229",
   rate_limiter=rate_limiter
)
```

## 带上你自己的模型¶
如果您所需的 LLM 未得到 LangChain 的官方支持，请考虑以下选项：

- 实现自定义 LangChain 聊天模型：创建符合LangChain 聊天模型接口的模型。这可以完全兼容 LangGraph 的代理和工作流，但需要了解 LangChain 框架。

- 使用自定义流直接调用：通过添加自定义流逻辑直接使用您的模型StreamWriter。请参阅自定义流文档以获取指导。此方法适用于无需预建代理集成的自定义工作流。
