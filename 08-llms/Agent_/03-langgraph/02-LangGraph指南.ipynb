
# LangGraph指南¶
本节中的页面提供了以下主题的概念概述和操作方法：

## 代理开发¶
- [概述](https://langchain-ai.github.io/langgraph/agents/overview/)：使用预构建的组件来构建代理。
- [运行代理](https://langchain-ai.github.io/langgraph/agents/run_agents/)：通过提供输入、解释输出、启用流和控制执行限制来运行代理。

## LangGraph API
- [Graph API](https://langchain-ai.github.io/langgraph/concepts/low_level/)：使用 Graph API 以图形范式定义工作流程。
- [功能性 API](https://langchain-ai.github.io/langgraph/concepts/functional_api/)：使用功能性 API 以功能性范式构建工作流，而无需考虑图形结构。
- [运行时](https://langchain-ai.github.io/langgraph/concepts/pregel/)：Pregel 实现 LangGraph 的运行时，管理 LangGraph 应用程序的执行。

## 核心能力
这些功能在 LangGraph OSS 和 LangGraph 平台中均可用。

- [流式传输](https://langchain-ai.github.io/langgraph/concepts/streaming/)：从 LangGraph 图流式输出。
- [持久性](https://langchain-ai.github.io/langgraph/concepts/persistence/)：持久化 LangGraph 图的状态。
- [持久执行](https://langchain-ai.github.io/langgraph/concepts/durable_execution/)：保存图形执行中关键点的进度。
- [记忆](https://langchain-ai.github.io/langgraph/concepts/memory/)：记住有关以前互动的信息。
- [上下文](https://langchain-ai.github.io/langgraph/agents/context/)：将外部数据传递到 LangGraph 图以提供图执行的上下文。
- [模型](https://langchain-ai.github.io/langgraph/agents/models/)：将各种 LLM 集成到您的 LangGraph 应用程序中。
- [工具](https://langchain-ai.github.io/langgraph/concepts/tools/)：直接与外部系统接口。
- [人机交互](https://langchain-ai.github.io/langgraph/concepts/human_in_the_loop/)：在工作流程的任何时候暂停图表并等待人工输入。
- [时间旅行](https://langchain-ai.github.io/langgraph/concepts/time-travel/)：回到 LangGraph 图执行过程中的某个特定点。
- [子图](https://langchain-ai.github.io/langgraph/concepts/subgraphs/)：构建模块化图。
- [多代理](https://langchain-ai.github.io/langgraph/concepts/multi_agent/)：将复杂的工作流程分解为多个代理。
- [MCP](https://langchain-ai.github.io/langgraph/concepts/mcp/)：在 LangGraph 图中使用 MCP 服务器。
- [评估](https://langchain-ai.github.io/langgraph/agents/evals/)：使用 LangSmith 评估图表的性能。

## 仅限平台的功能
这些功能仅在LangGraph 平台中可用。

- [身份验证和访问控制](https://langchain-ai.github.io/langgraph/concepts/auth/)：验证并授权用户访问 LangGraph 图。
- [助手](https://langchain-ai.github.io/langgraph/concepts/assistants/)：构建可用于与 LangGraph 图交互的助手。
- [双重短信](https://langchain-ai.github.io/langgraph/concepts/double_texting/)：在 LangGraph 图中处理双重短信（在返回第一个响应之前的连续消息）。
- [Webhooks](https://langchain-ai.github.io/langgraph/cloud/concepts/webhooks/)：将 webhooks 发送到 LangGraph 图。
- [Cron 作业](https://langchain-ai.github.io/langgraph/cloud/concepts/cron_jobs/)：安排作业在特定时间运行。
- [服务器定制](https://langchain-ai.github.io/langgraph/how-tos/http/custom_lifespan/)：定制运行 LangGraph 图的服务器。
- [数据管理](https://langchain-ai.github.io/langgraph/cloud/concepts/data_storage_and_privacy/)：以 LangGraph 图表形式管理数据。
- [部署](https://langchain-ai.github.io/langgraph/concepts/deployment_options/)：将 LangGraph 图部署到服务器。
