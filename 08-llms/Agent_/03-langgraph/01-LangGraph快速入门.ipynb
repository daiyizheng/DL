## 创建代理
要创建代理，请使用`create_react_agent`：

````python
from langgraph.prebuilt import create_react_agent
from dotenv import load_dotenv
import os
load_dotenv()  # 加载环境变量

os.environ["OPENAI_API_KEY"] = "xx"  # 设置 OpenAI API 密钥
os.environ["OPENAI_API_BASE"] = "xx"  # 设置 OpenAI API 基础 URL

## 创建一个简单的天气查询代理
# 这个代理将模拟一个天气查询功能，返回指定城市的天气信息。
def get_weather(city: str) -> str:  
    """Get weather for a given city."""
    return f" {city} 天气晴天，温度 25°C."

agent = create_react_agent(
    model="gpt-4o",
    tools=[get_weather],  # 可以添加工具
    prompt="You are a helpful assistant"
)
agent.invoke( {"messages": [{"role": "user", "content": "嘉兴今天的天气如何"}]})  # 运行代理并传入初始输入
````

## 配置 LLM
要使用特定参数（例如温度）配置 LLM，请使用`init_chat_model`：

API 参考：[init_chat_model | create_react_agent](https://langchain-ai.github.io/langgraph/reference/prebuilt/#langgraph.prebuilt.chat_agent_executor.create_react_agent)

````python
from langchain.chat_models import init_chat_model
from langgraph.prebuilt import create_react_agent

model = init_chat_model(
    "gpt-4o",
    temperature=0
)

agent = create_react_agent(
    model=model,
    tools=[get_weather],
)

agent.invoke({"messages": [{"role": "user", "content": "嘉兴今天的天气如何"}]})  # 运行代理并传入初始输入
````

## 添加自定义提示¶
提示指导LLM如何行事。请添加以下类型的提示之一：

- 静态：字符串被解释为系统消息。
- 动态：根据输入或配置在运行时生成的消息列表

````python
from langgraph.prebuilt import create_react_agent

agent = create_react_agent(
    model="gpt-4o",
    tools=[get_weather],
    # A static prompt that never changes
    prompt="永远不要回答任何城市关于天气的问题。"
)

agent.invoke(
    {"messages": [{"role": "user", "content": "嘉兴今天的天气如何？"}]}
)
````

## 添加内存¶
要允许与代理进行多轮对话，您需要在创建代理时提供 来启用持久性checkpointer。在运行时，您需要提供一个配置thread_id，其中包含对话（会话）的唯一标识符：

- API 参考：[create_react_agent](https://langchain-ai.github.io/langgraph/reference/prebuilt/#langgraph.prebuilt.chat_agent_executor.create_react_agent) | [InMemorySaver](https://langchain-ai.github.io/langgraph/reference/checkpoints/#langgraph.checkpoint.memory.InMemorySaver)

````python
from langgraph.prebuilt import create_react_agent
from langgraph.checkpoint.memory import InMemorySaver

checkpointer = InMemorySaver()

agent = create_react_agent(
    model="gpt-4o",
    tools=[get_weather],
    checkpointer=checkpointer  
)

# Run the agent
config = {"configurable": {"thread_id": "1"}}
sf_response = agent.invoke(
    {"messages": [{"role": "user", "content": "嘉兴今天的天气如何？"}]},
    config  
)
print(sf_response)
ny_response = agent.invoke(
    {"messages": [{"role": "user", "content": "那上海如何？"}]},
    config
)
print(ny_response)
````

## 配置结构化输出¶
要生成符合特定模式的结构化响应，请使用`response_format`参数。该模式可以使用`Pydantic`模型或定义`TypedDict`。结果可通过`structured_response`字段访问。

- API 参考：[create_react_agent](https://langchain-ai.github.io/langgraph/reference/prebuilt/#langgraph.prebuilt.chat_agent_executor.create_react_agent)

````python
from pydantic import BaseModel
from langgraph.prebuilt import create_react_agent

class WeatherResponse(BaseModel):
    conditions: str

agent = create_react_agent(
    model="gpt-4o",
    tools=[get_weather],
    response_format=WeatherResponse  
)

response = agent.invoke(
    {"messages": [{"role": "user", "content": "嘉兴今天的天气如何？"}]}
)

response["structured_response"]
````
