# Agent 智能体设计模式

## 增强型LLM(augmented LLM)-RAG
Agent 系统的基本构建模块是 LLM，并通过检索、工具和记忆等功能进行增强。我们当前的模型可以主动使用这些功能——生成自己的搜索查询、选择合适的工具并确定要保留的信息。

```python
from openai import OpenAI
from datetime import datetime
import json
from typing import List, Dict, Callable
import os
import re
from dotenv import load_dotenv
load_dotenv("/Users/a1-6/Documents/projects/DL/.env")
client = OpenAI(
    api_key=os.getenv("DASHSCOPE_API_KEY"),
    base_url="https://dashscope.aliyuncs.com/compatible-mode/v1",
)
```

```python
llm_name = llm_name = "qwen-plus"
def call_llm(user_prompt, system_prompt=""):
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user","content": user_prompt}
        ]
    response = client.chat.completions.create(
            model= llm_name,
            messages = messages)
    return(response.choices[0].message.content)
```

```python
content = """
白术具有增强体质、健脾益气、利尿消肿、固表止汗、安胎的功效。作用是提高免疫力、改善脾胃虚弱、消除水肿、治疗表虚自汗、预防胎动不安。服用后若出现不适症状，建议及时就医。

1、增强体质：
白术中所含有的白术多糖能够调节人体免疫功能，提高免疫力，对于免疫力低下所引起的疾病具有一定的预防效果。

2、健脾益气：
白术归脾、胃经，具有健脾益气的功效，对于脾胃虚弱所引起的食欲不振、消化不良等症状具有一定的缓解作用。

3、利尿消肿：
白术具有一定的利尿作用，能够帮助体内多余水分的排出，因此可以有效缓解水肿、尿频等症状。

4、固表止汗：
白术甘、苦，性温，适量服用后能够固护肌表，防止汗液过多流失，对于自汗、盗汗等症状有一定的改善作用。

5、安胎：
白术是常用于保胎的一味中药材，可以治疗胎动不安、妊娠呕吐等症状。

在服药期间建议患者调整饮食习惯，以清淡、易消化的食物为主，如小米粥、鸡蛋羹等，这样有利于疾病的恢复。

"""

user_query = "白术功效是什么？"
```

```python
call_llm(user_prompt=user_query, system_prompt=content)
```
