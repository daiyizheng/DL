# 线程池
Starlette 在几种情况下使用线程池来避免阻塞事件循环：

- def当您使用而不是创建同步端点时`async def`
- 使用以下方式提供文件时`FileResponse`
- 使用以下方式处理文件上传`UploadFile`
- 使用以下方式运行同步后台任务`BackgroundTask`
- 还有一些可能没有记录的场景......

`Starlette` 会在线程池中运行您的代码，以避免阻塞事件循环。这不仅适用于您创建的端点函数和后台任务，也适用于 `Starlette` 内部代码。

更准确地说，`Starlette` 使用`anyio.to_thread.run_sync`同步代码运行。


## 并发限制
默认线程池大小仅为 `40 `个令牌。这意味着只能同时运行 **40 个线程**。此限制与其他库相同：例如，FastAPI 也用于anyio运行同步依赖项，这也会耗尽线程容量。

如果需要运行更多线程，可以增加令牌的数量：

```python
import anyio.to_thread

limiter = anyio.to_thread.current_default_thread_limiter()
limiter.total_tokens = 100
```

上述代码将把令牌数量增加到100。

增加线程数量可能会对性能和内存产生影响，因此在这样做时要小心。
